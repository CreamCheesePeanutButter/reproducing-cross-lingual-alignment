# Hugging face and NDIF login

```python
import os
from google.colab import userdata
from nnsight import login, whoami

# Load your secret API keys safely
os.environ["HF_TOKEN"] = userdata.get('HF_TOKEN')
os.environ["NDIF_API_KEY"] = userdata.get('NDIF_API_KEY')

# Create necessary folders inside Colab to save your experiment results
os.environ["HF_HOME"] = "/content/hf_cache"
os.makedirs("/content/accuracy", exist_ok=True)
os.makedirs("/content/patching", exist_ok=True)

# Test your NDIF connection to ensure your key works instantly
try:
    login(os.environ["NDIF_API_KEY"])
    whoami()
    print("🚀 NDIF Connection Successful! You are ready to run experiments.")
except Exception as e:
    print(f"❌ Connection failed. Check your API key. Error: {e}")
```
